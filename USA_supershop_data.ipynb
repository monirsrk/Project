```python
sns.countplot(x='Category',data=df[df['Category']=='Furniture'],hue='Sub-Category')
```

```python
sns.catplot(x='Sub-Category',data=df[df['Category']=='Office Supplies'],hue='Sub-Category',order=df[df['Category']=='Office Supplies']['Sub-Category'].value_counts().index,kind='count', height=8, aspect=2)
plt.xticks(rotation=90)
plt.title('Count of Sub-Categories within Office Supplies')
plt.show()
```

```python
sns.catplot(x='Sub-Category', data=df[df['Category']=='Office Supplies'], kind='count', height=8, aspect=2, order=df[df['Category']=='Office Supplies']['Sub-Category'].value_counts().index)
plt.xticks(rotation=90)
plt.title('Count of Sub-Categories within Office Supplies')
plt.show()
```
